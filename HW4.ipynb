```python
# 安裝必要套件
!pip install pycaret optuna seaborn pandas matplotlib --upgrade --no-cache-dir

# 匯入模組
from pycaret.classification import *
import optuna
import pandas as pd
import seaborn as sns

# 讀取 Titanic 資料集
data = sns.load_dataset('titanic')

# **1. 特徵工程**
data = data.drop(['deck', 'embark_town', 'alive', 'who', 'class', 'adult_male'], axis=1)
data['age'] = data['age'].fillna(data['age'].median())
data['embarked'] = data['embarked'].fillna(data['embarked'].mode()[0])
data['sex'] = data['sex'].astype('category')
data['embarked'] = data['embarked'].astype('category')
data['age_fare'] = data['age'] * data['fare']

# 初始化 PyCaret 環境
clf_setup = setup(data=data,
                  target='survived',
                  session_id=123,
                  normalize=True,
                  categorical_features=['sex', 'embarked'])

# **2. 模型選擇**
best_model = compare_models(sort='Accuracy')

# **3. 超參數優化 (使用 Optuna)**
def objective(trial):
    params = {
        'learning_rate': trial.suggest_loguniform('learning_rate', 0.01, 0.2),
        'max_depth': trial.suggest_int('max_depth', 3, 10),
        'n_estimators': trial.suggest_int('n_estimators', 50, 200),
        'num_leaves': trial.suggest_int('num_leaves', 20, 50),
        'subsample': trial.suggest_uniform('subsample', 0.5, 1.0),
        'colsample_bytree': trial.suggest_uniform('colsample_bytree', 0.5, 1.0),
    }
    # 使用參數構建模型
    model = create_model('lightgbm', **params)
    # 調整模型
    tuned_model = tune_model(model, optimize='Accuracy')
    # 獲取準確率
    results = pull()
    return results.loc[results.index[0], 'Accuracy']

# 使用 Optuna 進行超參數搜尋
study = optuna.create_study(direction='maximize')
study.optimize(objective, n_trials=30)

# 獲取最佳參數
print("最佳參數：", study.best_params)

# 使用最佳參數構建最終模型
final_model = create_model('lightgbm', **study.best_params)

# **模型評估與保存**
plot_model(final_model, plot='auc')
plot_model(final_model, plot='confusion_matrix')
save_model(final_model, 'optimized_lightgbm_model')

# 下載模型
from google.colab import files
files.download('optimized_lightgbm_model.pkl')
```
